```python
from ultralytics import YOLO

# Load the YOLO11 model
model = YOLO("models/DetectionYOLO11n.pt", task="detect")

# Export the model to CoreML format
model.export(format="coreml", nms=True, device="mps")  # creates 'yolo11n.mlpackage'
```

```python

# Load the exported CoreML model
coreml_model = YOLO("models/yolo11m.mlpackage", task="detect")

# Run inference
results = coreml_model("https://ultralytics.com/images/bus.jpg")
```
